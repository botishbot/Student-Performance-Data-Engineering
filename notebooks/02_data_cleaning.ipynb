```python
# Import required libraries
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Load the original dataset
file_path = "../data/raw/DTS_202_Dataset.xlsx"

df = pd.read_excel(file_path)

# Create a separate working copy
clean_df = df.copy()

print("Dataset loaded successfully!")
print("Original dataset shape:", df.shape)
print("Working dataset shape:", clean_df.shape)
```

```python
# Check missing values
print("Missing values:")
print(clean_df.isnull().sum())

print("\nTotal missing values:", clean_df.isnull().sum().sum())

# Check duplicate records
print("\nDuplicate records:", clean_df.duplicated().sum())
```

```python
# Record the initial dataset size
initial_rows = len(clean_df)

# Remove duplicate records (if any)
clean_df = clean_df.drop_duplicates().copy()

# Check whether any missing values remain
remaining_missing = clean_df.isnull().sum().sum()

# Display cleaning results
print("DATA CLEANING SUMMARY")
print("---------------------")
print("Initial records:", initial_rows)
print("Records after duplicate removal:", len(clean_df))
print("Duplicate records removed:", initial_rows - len(clean_df))
print("Remaining missing values:", remaining_missing)
print("Final dataset shape:", clean_df.shape)
```

## Outlier Detection and Handling

The Interquartile Range (IQR) method and boxplots
were used to investigate potential outliers.

The absences column contained 15 potential outliers,
with values exceeding the upper IQR boundary of 20.

The flagged records were inspected. Since there was
insufficient evidence that these observations were
incorrect, all 15 records were retained.

No outliers were detected in the G3 final-grade column.

Therefore, no outlier removal or capping was performed.

```python
# Save the cleaned dataset
output_path = "../data/processed/student_performance_cleaned.csv"

clean_df.to_csv(output_path, index=False)

print("Cleaned dataset saved successfully!")
print("Location:", output_path)
```

```python
# Reload the saved cleaned dataset
verification_df = pd.read_csv(output_path)

print("Saved dataset verification")
print("Shape:", verification_df.shape)
print("Missing values:", verification_df.isnull().sum().sum())
print("Duplicate records:", verification_df.duplicated().sum())
```
